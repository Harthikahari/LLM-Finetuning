# Text2SQL Inference Demo

Interactive demo for generating SQL queries.

```python
import sys
sys.path.append('..')

from text2sql.models.model_factory import create_model
from text2sql.inference.predictor import Text2SQLPredictor
```

```python
# Load model
model = create_model('./outputs', training_method='lora')
predictor = Text2SQLPredictor(model, model.tokenizer)
```

```python
# Generate SQL
question = "What are the names of all employees?"
queries = predictor.predict(question)
print(f"Question: {question}")
print(f"SQL: {queries[0]}")
```
