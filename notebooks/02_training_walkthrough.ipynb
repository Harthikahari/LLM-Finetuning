# Text2SQL Training Walkthrough

Step-by-step guide to fine-tuning a Text2SQL model.

```python
import sys
sys.path.append('..')

from text2sql.models.model_factory import create_model
from text2sql.data.datasets import load_dataset, Text2SQLDataset
from text2sql.training.trainer import create_trainer
```

## 1. Create Model

```python
model = create_model(
    model_name='codellama-7b',
    training_method='qlora',
    lora_r=16,
    lora_alpha=32
)
```

## 2. Load Data

```python
train_examples = load_dataset('spider', split='train')
eval_examples = load_dataset('spider', split='validation')
```

## 3. Train Model

```python
# Create datasets
train_dataset = Text2SQLDataset(train_examples, model.tokenizer)
eval_dataset = Text2SQLDataset(eval_examples, model.tokenizer)

# Create trainer
trainer = create_trainer(
    model=model.model,
    tokenizer=model.tokenizer,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    num_train_epochs=1
)

# Train
trainer.train()
```
