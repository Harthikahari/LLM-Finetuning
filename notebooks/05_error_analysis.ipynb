# Error Analysis

Analyze model errors and identify improvement areas.

```python
import sys
sys.path.append('..')

import pandas as pd
from text2sql.models.model_factory import create_model
from text2sql.data.datasets import load_dataset
```

```python
# Load model and data
model = create_model('./outputs', training_method='lora')
test_data = load_dataset('spider', split='test')[:100]  # First 100 examples
```

```python
# Analyze errors
errors = []
for example in test_data:
    pred = model.generate(example.question)[0]
    if pred != example.query:
        errors.append({
            'question': example.question,
            'expected': example.query,
            'predicted': pred
        })

df = pd.DataFrame(errors)
print(f"Error rate: {len(errors)/len(test_data)*100:.2f}%")
df.head()
```
